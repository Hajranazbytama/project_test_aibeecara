### Melakukan Labeling menggunakan ROBERTA

```python
# 1. (jika belum) install dependencies:
#    pip install pandas transformers torch

import pandas as pd
from transformers import pipeline

# 2. Load data
df = pd.read_csv('data_stemm.csv')

# 3. Inisialisasi sentiment pipeline (Indonesia)
sent_id = pipeline(
    "sentiment-analysis",
    model="w11wo/indonesian-roberta-base-sentiment-classifier",
    tokenizer="w11wo/indonesian-roberta-base-sentiment-classifier",
    device=-1  # pakai GPU jika ada, ganti ke -1 untuk CPU
)

# 4. Fungsi untuk memanggil pipeline dan map label
def get_sentiment_id(text):
    # batasi panjang agar tidak error
    txt = text[:512] if isinstance(text, str) else ""
    out = sent_id(txt)[0]             # {'label': 'POSITIVE'/'NEGATIVE'/'NEUTRAL', 'score': ...}
    lbl = out['label'].lower()        # jadi 'positive'/'negative'/'neutral'
    # kembali dalam bahasa sederhana
    if 'neg' in lbl: return 'negative'
    if 'pos' in lbl: return 'positive'
    return 'neutral'

# 5. Terapkan ke semua baris
#    Abaikan NaN atau non-string
df['sentiment'] = df['stemmed'].fillna("").astype(str).apply(get_sentiment_id)

df.head()
# 6. Simpan hasil ke CSV baru
df.to_csv('data_stemm_labeled.csv', index=False)

print("Selesai labeling, file `data_stemm_labeled.csv` sudah ter-generate.")
```
